## 1. Ranked actions + reason codes

We translate raw model probabilities into five concrete operational playbooks:
1. `refresh_and_review_ctr` (22.2%): High SERP visibility but below-average CTR -> Optimize snippet, title, meta description.
2. `refresh` (27.4%): Established, decaying content -> Update copy, fresh data, internal links.
3. `refresh_and_review_engagement` (6.6%): Low on-page dwell / scroll depth -> Redesign UX, visual formatting.
4. `expand_and_refresh` (0.3%): Thin content in high-competition space -> Expand into comprehensive pillar guide.
5. `monitor` (43.6%): Stable, low-risk content -> Protect against unnecessary edits.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load scored outputs from reference pipeline
queue_path = Path('outputs/refresh_queue.csv')
if not queue_path.exists():
    queue_path = Path('outputs/refresh_queue_sample.csv')

df_queue = pd.read_csv(queue_path)
print(f"Loaded queue of {len(df_queue):,} scored pages.")
action_counts = df_queue['suggested_action'].value_counts()
display(pd.DataFrame({'Count': action_counts, 'Share': df_queue['suggested_action'].value_counts(normalize=True).map('{:.1%}'.format)}))
```

## 3. Human review + the no-go list

**Human Review Protocol:**
1. Check current live URL status and ensure no technical crawl errors exist.
2. Verify search intent: has user intent changed from informational to transactional?
3. Formulate specific editorial hypothesis prior to editing.

**The No-Go List (Refusal to Automate):**
- Never auto-rewrite pages using generative AI without editorial oversight.
- Never modify pages currently on an organic growth streak (`monitor` status).
- Never redirect or prune URLs with valuable referral backlink equity without manual audit.

```python
print("Human-in-the-loop safeguards and no-go constraints implemented.")
```

## 4. Monitoring / retrain triggers

**Retrain Triggers:**
1. **Base Rate Drift:** If portfolio decline rate shifts by > 15% month-over-month.
2. **Feature Drift (PSI):** If Population Stability Index (PSI) on `days_with_impressions` or `avg_position` exceeds 0.20.
3. **Quarterly Refresh:** Re-train models every 90 days incorporating the latest snapshot.

```python
print("Monitoring harness configured: PSI thresholds set at 0.10 (warning) and 0.20 (critical drift).")
```
